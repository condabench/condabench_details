```python
df=pd.read_csv('/kaggle/input/world-happiness/2018.csv')
df.columns
```

```python
col1=df['Social support']
col2=df['Freedom to make life choices']
max1=col1.max()
max2=col2.max()
min1=col1.min()
min2=col2.min()
mid_range1=(max1+min1)/2
mid_range2=(max2+min2)/2
print("Mid range of social support",mid_range1)
print("Mid range of Freedom to make life choices ",mid_range2)
```

```python
variance=df[ 'Freedom to make life choices'].var()
std_dev=df[ 'Freedom to make life choices'].std()
print(variance)
print(std_dev)
```

```python
from scipy import stats
z_scores = stats.zscore(df['GDP per capita'])
threshold = 3
outliers = (z_scores > threshold) | (z_scores < -threshold)
df_cleaned = df[outliers]
mean_gdp = df_cleaned['GDP per capita'].mean()
variance_gdp = df_cleaned['GDP per capita'].var()
print("Mean of GDP per capita without outliers",mean_gdp)
print("Variance of GDP per capita without outliers",variance_gdp)
```
