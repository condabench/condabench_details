```python
import plotly.graph_objects as go

# Example using a dropdown to select different metrics
fig = go.Figure()

# Happiness score trace
fig.add_trace(go.Choropleth(
    locations=df19['Country or region'],
    locationmode='country names',
    z=df19['Score'],
    colorscale='Viridis',
    colorbar_title='Happiness Score',
    visible=True
))

# Add more traces for other metrics like GDP per capita or Social support
# ...

# Add dropdown menu
fig.update_layout(
    updatemenus=[{
        'buttons': [
            {
                'label': 'Happiness Score',
                'method': 'update',
                'args': [{'visible': [True, False, False]}]  # Show happiness score trace
            },
            {
                'label': 'GDP per capita',
                'method': 'update',
                'args': [{'visible': [False, True, False]}]  # Show GDP trace
            },
            # Add buttons for other metrics
        ],
        'direction': 'down',
        'showactive': True
    }]
)

fig.show()
```

```python
from sklearn.decomposition import PCA
import numpy as np

# Assuming `df` has columns for the factors you're interested in
X = df19[['GDP per capita', 'Social support', 'Healthy life expectancy', 'Freedom to make life choices', 'Generosity', 'Perceptions of corruption']]
pca = PCA(n_components=2)
principal_components = pca.fit_transform(X)

# Add PCA results to DataFrame
df_pca = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
df_pca['Country'] = df19['Country or region']

# Plot PCA results
fig = px.scatter(df_pca, x='PC1', y='PC2', text='Country', title='PCA of Happiness Factors')
fig.show()
```

```python
import statsmodels.api as sm

# Prepare data
X = df19[['GDP per capita', 'Social support', 'Healthy life expectancy', 'Freedom to make life choices', 'Generosity', 'Perceptions of corruption']]
y = df19['Score']

# Add constant term
X = sm.add_constant(X)

# Fit model
model = sm.OLS(y, X).fit()

# Print summary
print(model.summary())
```

```python
from sklearn.cluster import KMeans

# Prepare data
X = df19[['GDP per capita', 'Social support', 'Healthy life expectancy', 'Freedom to make life choices', 'Generosity', 'Perceptions of corruption']]
kmeans = KMeans(n_clusters=5)  # Choose the number of clusters
clusters = kmeans.fit_predict(X)

# Add cluster information to DataFrame
df19['Cluster'] = clusters

# Plot clustering results
fig = px.scatter(df19, x='GDP per capita', y='Score', color='Cluster', title='Clustering of Countries by Happiness Factors')
fig.show()
```
