```python
duration= nf['duration'].value_counts().reset_index().sort_values(by='count',ascending=False).head(10)
duration
```

```python
def convert_duration(duration):
    if 'min' in duration:
        minutes = int(duration.split()[0])
        if(minutes>30 and minutes <= 75):
            hours = 1
        elif(minutes>75 and minutes <105 ):
            hours= 1.5
        elif(minutes>=105 and minutes<150):
            hours=2
        else:
            hours= 3
        
        return str(hours)+' hours'
    elif 'Season' in duration:
        season= int(duration.split()[0])
        if season not in (1,2,3):
            return 'more than 3 ses'
        return duration # Keep seasons as they are
    return np.nan
```

```python
nf['dura']=[convert_duration(d) for d in nf['duration']]
nf['dura']
```

```python
plt.figure(figsize=(15,7))
sns.histplot(x='dura', data=nf,hue='type')
plt.xlabel('Duration')
plt.ylabel('Movies count')

```
